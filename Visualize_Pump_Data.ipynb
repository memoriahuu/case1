## Setup

```python
# Import Libraries 

import numpy as np                                          # Import Numpy for data manipulation
import matplotlib.pyplot as plt                             # Import Matplotlib for data visualization
```

## Import Dataset

```python
#Import data from one file 

pump1 = np.genfromtxt("pump_data/Data - pump 1.csv", delimiter=",")
pump1
```

```python
print(pump1.shape)                                          # does the size of the data fit with what we expect? 
```

```python
pump1 = np.delete(pump1, 0, axis=0)                         # delete header row which is text (nan) 
print(pump1.shape)
```

```python
# Create Q, H, rpm Data Arrays  

Q = pump1[:, 0]                                             # Q is the first column of the data
H = pump1[:, 1]                                             # H is the second column of the data
rpm = pump1[:, 3]                                           # rpm is the fourth column of the data
```

## Visualize Data

```python
# Visualize Data 

plt.scatter(Q, H, c=rpm)
plt.show() 
```

```python
# Select Q & H for a given range of rpm 

Q1 = Q[(rpm > 0.58) & (rpm < 0.61)]                             # Select Q values for rpm between 0.58 and 0.61
H1 = H[(rpm > 0.58) & (rpm < 0.61)]                             # Select H values for rpm between 0.58 and 0.61

plt.scatter(Q1, H1)
plt.show()
```

```python
# 3D scatter plot 

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')         

ax.scatter(Q, rpm, H, zdir='z', c='red')

ax.set_xlabel('Q [m3/s]')
ax.set_ylabel('RPM')
ax.set_zlabel('H [m]')

plt.show()
```

## Ideal Models 

```python
# define "ideal" model / functions 

def pump_model_QH(Q,rpm):                                           # function to model head as a function of flow rate and rpm, provided by Grundfos 
    QH_coeff = [-0.580818, 1.65095, -1.4245, 0.105362, 0.104778, 0.249879]
    nref = .5
    QH_func = np.poly1d(QH_coeff)                                   # polynomial function of degree 5 (based on number of parameters passed)
    H = QH_func(Q)*(rpm/nref)**2                                    # scaling law for head with respect to rpm 
    return H

def pump_model_QP(Q,rpm):
    QP_coeff = [-0.263138, 0.0574016, 0.517813, 0.0929127]
    nref = .5
    QP_func = np.poly1d(QP_coeff)
    P = QP_func(Q)*(rpm/nref)**2
    return P
```

```python
# Ideal QH curve at fixed rpm 

rpm_fixed = 0.6 
Q2 = np.linspace(0, 1, 1000)    
H2 = pump_model_QH(Q2, rpm_fixed)

plt.scatter(Q2, H2)
plt.show()
```

```python
# create ideal H values given our dataset of Q and rpm values 

H1 = pump_model_QH(Q, rpm)

print(H1.size)
```

```python
# Visualize Data 

plt.scatter(Q, H1, c=rpm)
plt.show() 
```

```python
# 3D scatter plot 

fig2 = plt.figure()
ax = fig2.add_subplot(111, projection='3d')

ax.scatter(Q, rpm, H1, zdir='z', c='red')

ax.set_xlabel('Q [m3]')
ax.set_ylabel('RPM')
ax.set_zlabel('H Ideal [m]')

plt.show()
```

## Ideal vs Measured Data

```python
# Compare real data with ideal curve

QR = Q[(rpm > 0.58) & (rpm < 0.60)]     # Select Q & H for a given range of rpm from measured data
HR = H[(rpm > 0.58) & (rpm < 0.60)]

rpm_fixed = 0.59                        # Ideal QH curve at fixed rpm
QI = np.linspace(0, 1, 1000)    
HI = pump_model_QH(QI, rpm_fixed)

plt.scatter(QR, HR)                     # Real data
plt.scatter(QI, HI)                     # Ideal curve   
plt.show()
```

```python
# Plot real data and ideal curve for all rpm values

plt.scatter(Q, H, c='r')            # Real data in red
plt.scatter(Q, H1, c='b')           # Ideal curve in blue 
plt.show() 
```

## Residuals

```python
# H residual is the difference between the measured (H) and ideal value (H1) 

HR = H - H1
```

```python
# Visualize Data 

plt.scatter(Q, HR, c=rpm)
plt.show() 
```

```python
# 3D scatter plot 

fig2 = plt.figure()
ax = fig2.add_subplot(111, projection='3d')

ax.scatter(Q, rpm, HR, zdir='z', c='red')

ax.set_xlabel('Q [m3]')
ax.set_ylabel('RPM')
ax.set_zlabel('HI deal [m]')

plt.tight_layout()

plt.show()
```
